```python
df['Profitability Flag'] = np.where(
    df['Order Profit Per Order'] > 0,
    'Profit',
    np.where(
        df['Order Profit Per Order'] < 0,
        'Loss',
        'Break-even'
    )
)

df['Profitability Flag'].value_counts()                      
```

```python
#Visualization of profitability

profit_counts = df['Profitability Flag'].value_counts(normalize=True)*100     # 100% accuracy hota h normalization se
profit_counts.plot(kind='pie' , colors = ['Maroon', 'green','Blue'])
plt.ylabel('')
plt.title('Profitability Distribution(%)')
plt.show()
```

```python
# Bussiness KPI


# KPI Values
total_sales = df['Order Item Total'].sum()
total_profit = df['Order Profit Per Order'].sum()
total_orders = df['Delivery Status'].nunique()

# Create figure
fig, ax = plt.subplots(figsize=(12,4))

ax.axis('off')

# KPI Text
plt.text(0.1, 0.5, f'Order Item Total\n{total_sales:,.0f}',
         fontsize=18, weight='bold',
         bbox=dict(facecolor='skyblue', alpha=0.5))

plt.text(0.4, 0.5, f'Total Profit\n{total_profit:,.0f}',
         fontsize=18, weight='bold',
         bbox=dict(facecolor='lightgreen', alpha=0.5))

plt.text(0.7, 0.5, f'Delivery Status\n{total_orders}',
         fontsize=18, weight='bold',
         bbox=dict(facecolor='orange', alpha=0.5))

plt.show()
```
